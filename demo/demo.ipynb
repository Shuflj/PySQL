# SQL cell variables demo

1. Add `demo/shop.db` as a SQLite data source in the Database tool window.
2. Run the Python cell below.
3. Add a SQL cell, pick the `shop.db` data source, and paste one of the queries from the last cell.

````python
import datetime
import pandas

user_id = 42
min_price = 50
products = ["Keyboard", "Monitor", "O'Reilly book"]
since = datetime.date(2026, 10, 1)
table = "orders"
app_values = [("web", 1), ("ios", 2)]
````

````python
%%sql
WITH apps(app_name, app) AS (VALUES {{ app_values }}) SELECT * FROM apps
````

Queries to paste into a SQL cell:

```sql
SELECT * FROM orders WHERE user_id = {{ user_id }} AND price > {{ min_price }}

SELECT * FROM {{ table | raw }} WHERE product IN {{ products }} AND ordered_on >= {{ since }}

SELECT * FROM orders WHERE product LIKE '%{{ products[2].split()[0] }}%'

WITH apps(app_name, app) AS (VALUES {{ app_values }}) SELECT * FROM apps
```
